```python
import re

FILENAME_RE = re.compile(
    r"""
    ^
    (?P<model_family>[A-Za-z]+)_
    p(?P<teff>\d+)
    g(?P<logg>-?\d+(?:\.\d+)?)
    z(?P<feh>-?\d+(?:\.\d+)?)
    t(?P<vturb>-?\d+(?:\.\d+)?)
    _
    a(?P<alpha>-?\d+(?:\.\d+)?)
    c(?P<cfe>-?\d+(?:\.\d+)?)
    n(?P<nfe>-?\d+(?:\.\d+)?)
    o(?P<ofe>-?\d+(?:\.\d+)?)
    r(?P<rfe>-?\d+(?:\.\d+)?)
    s(?P<sfe>-?\d+(?:\.\d+)?)
    _
    (?P<domain>[A-Z]+)
    (?:\.spec)?     # Optional '.spec' extension
    $
    """,
    re.VERBOSE,
)


def parse_spectrum_filename(filename: str) -> dict:
    # Remove any trailing file extensions, if present
    base = filename
    for ext in [".fits", ".fit", ".gz", ".bz2", ".txt", ".dat"]:
        if base.lower().endswith(ext):
            base = base[: -len(ext)]
    m = FILENAME_RE.match(base)
    if not m:
        raise ValueError(f"Unrecognized filename format: {filename}")

    d = m.groupdict()

    # Cast numerics
    for k in [
        "teff", "logg", "feh", "vturb",
        "alpha", "cfe", "nfe", "ofe", "rfe", "sfe"
    ]:
        d[k] = float(d[k]) if "." in d[k] or "-" in d[k] else int(d[k])

    return d
```

```python
from pathlib import Path
import json
import numpy as np
import zarr
from zarr.codecs import ZstdCodec
from astropy.io import fits
from astropy.table import Table
from astropy.wcs import WCS

def convert_fits_to_zarr_shard(
    fits_path,
    zarr_out,
    *,
    hdu_index=1,
    flux_col="flux",
    norm_flux_col="normalized flux",
    wave_col="wavelength",
    batch_chunk=128,
    compression_level=3,
    cast_float32=True,
):
    """
    Convert a single FITS file into a single Zarr shard (Zarr v3 compliant).

    Stored arrays:
      - flux        (N_spec, N_lambda)
      - flux_norm   (N_spec, N_lambda)
      - wavelength  (N_lambda,)

    Parameters
    ----------
    fits_path : str or Path
        Input FITS file.
    zarr_out : str or Path
        Output Zarr directory.
    hdu_index : int
        HDU index containing the spectra table.
    flux_col : str
        Column name for raw flux.
    norm_flux_col : str
        Column name for normalized flux.
    wave_col : str or None
        Column name for wavelength array.
        If None, wavelength is reconstructed from WCS.
    batch_chunk : int
        Chunk size along spectrum axis (axis=0).
    compression_level : int
        Zstd compression level (1–3 for training, 5–7 for archival).
    cast_float32 : bool
        Cast flux arrays to float32 (recommended for ML).
    """

    fits_path = Path(fits_path)
    zarr_out = Path(zarr_out)

    # -------------------------
    # Read FITS
    # -------------------------
    with fits.open(fits_path, memmap=True) as hdul:
        hdu = hdul[hdu_index]

        if not isinstance(hdu, fits.BinTableHDU):
            raise ValueError(
                f"HDU {hdu_index} is not a BinTableHDU "
                "(required for flux + normalized flux)."
            )

        header = dict(hdu.header)
        table = Table(hdu.data)

        flux = np.asarray(table[flux_col])
        flux_norm = np.asarray(table[norm_flux_col])

        if wave_col is not None:
            wavelength = np.asarray(table[wave_col])
        else:
            wcs = WCS(hdu.header)
            n_lambda = flux.shape[1]
            wavelength = wcs.pixel_to_world(
                np.arange(n_lambda), 0
            )[0].value

    # -------------------------
    # Validate shapes
    # -------------------------
    if flux.shape != flux_norm.shape:
        raise ValueError("FLUX and FLUX_NORM shapes do not match")

    # Detect data dimensions and set chunk shapes correctly
    if flux.ndim == 1:  # Single spectrum, shape (N_lambda,)
        n_spec = 1
        n_lambda = flux.shape[0]
        flux = flux.reshape(1, n_lambda)
        flux_norm = flux_norm.reshape(1, n_lambda)
    elif flux.ndim == 2: # (N_spec, N_lambda)
        n_spec, n_lambda = flux.shape
    else:
        raise ValueError("Flux array must be 1D or 2D, got shape %r" % (flux.shape,))

    # Wavelength should always be (N_lambda,)
    if wavelength.ndim == 2 and wavelength.shape[0] == 1:
        wavelength = wavelength[0]
    elif wavelength.ndim == 2 and wavelength.shape[1] == 1:
        wavelength = wavelength[:,0]
    if wavelength.shape != (n_lambda,):
        if wavelength.shape == (n_spec, n_lambda):
            # Same wavelength replicated for every spectrum; just take first row
            wavelength = wavelength[0]
        elif wavelength.shape[0] == n_lambda:
            pass
        else:
            raise ValueError(
                f"Wavelength shape mismatch: expected ({n_lambda},), got {wavelength.shape}"
            )

    # -------------------------
    # Dtype handling
    # -------------------------
    if cast_float32:
        flux = flux.astype("f4", copy=False)
        flux_norm = flux_norm.astype("f4", copy=False)
        wavelength = wavelength.astype("f8", copy=False)

    # -------------------------
    # Zarr v3 setup
    # -------------------------
    codec = ZstdCodec(level=compression_level)
    compressors = [codec]

    chunk_shape_2d = (min(batch_chunk, n_spec), n_lambda)
    chunk_shape_1d = (n_lambda,)

    root = zarr.open_group(zarr_out, mode="w")

    # -------------------------
    # Write arrays (v3-safe)
    # -------------------------
    root.create_array(
        name="flux",
        data=flux,
        chunks=chunk_shape_2d,
        compressors=compressors,
    )

    root.create_array(
        name="flux_norm",
        data=flux_norm,
        chunks=chunk_shape_2d,
        compressors=compressors,
    )

    root.create_array(
        name="wavelength",
        data=wavelength,
        chunks=chunk_shape_1d,
        compressors=compressors,
    )

    # -------------------------
    # Metadata
    # -------------------------
    root.attrs.update(
        {
            "source_fits": fits_path.name,
            "hdu_index": hdu_index,
            "flux_col": flux_col,
            "norm_flux_col": norm_flux_col,
            "wave_col": wave_col,
            "compression": "zstd",
            "compression_level": compression_level,
            "dtype_flux": str(flux.dtype),
            "dtype_wavelength": str(wavelength.dtype),
            "header": json.dumps(header),
            "zarr_format": 3,
        }
    )
    
    root.attrs.update(parse_spectrum_filename(f[1].header["EXTNAME"]))

    return root
```
